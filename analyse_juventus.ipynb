```python
# ÉTAPE 1 : les données (millions d'euros), un élément par année
annees = ["2018/19", "2019/20", "2020/21", "2021/22", "2022/23"]
chiffre_affaires = [621.5, 573.4, 480.7, 443.4, 507.7]
charges_exploitation = [458.5, 417.0, 466.3, 468.3, 427.6]
resultat_net = [-39.9, -92.5, -226.8, -239.3, -123.7]
dette_nette = [463.5, 385.2, 389.2, 153.0, 339.9]
```

```python
# ÉTAPE 2 : calcul de l'EBITDA = chiffre d'affaires - charges d'exploitation
ebitda = []
for i in range(5):
    valeur = chiffre_affaires[i] - charges_exploitation[i]
    ebitda.append(valeur)
```

```python
# ÉTAPE 3 : calcul de la marge d'EBITDA en % = EBITDA / chiffre d'affaires * 100
marge = []
for i in range(5):
    m = ebitda[i] / chiffre_affaires[i] * 100
    marge.append(m)
```

```python
# ÉTAPE 4 : calcul de la croissance du chiffre d'affaires par rapport à l'année précédente
# (la première année n'a pas d'année précédente, donc je commence à 1)
croissance = []
for i in range(1, 5):
    c = (chiffre_affaires[i] - chiffre_affaires[i - 1]) / chiffre_affaires[i - 1] * 100
    croissance.append(c)
```

```python
# ÉTAPE 5 : affichage des résultats
for i in range(5):
    print("Année :", annees[i])
    print("  Chiffre d'affaires :", chiffre_affaires[i])
    print("  EBITDA :", round(ebitda[i], 1))
    print("  Marge d'EBITDA (%) :", round(marge[i], 1))
    if i > 0:
        print("  Croissance du CA (%) :", round(croissance[i - 1], 1))
    print("  Résultat net :", resultat_net[i])
    # le ratio dette/EBITDA n'a de sens que si l'EBITDA est positif
    if ebitda[i] > 0:
        print("  Dette/EBITDA :", round(dette_nette[i] / ebitda[i], 1))
    else:
        print("  Dette/EBITDA : non significatif (EBITDA négatif)")
    print()
```

```python
# ÉTAPE 6 : le graphique
import matplotlib.pyplot as plt

plt.plot(annees, ebitda, marker="o", label="EBITDA")
plt.plot(annees, resultat_net, marker="o", label="Résultat net")
plt.title("Juventus : EBITDA et résultat net (millions d'euros)")
plt.xlabel("Exercice")
plt.ylabel("Millions d'euros")
plt.legend()
plt.grid(True)
plt.show()
```
